# Project Title: Predict House Prices with Tree-Based Non-Linear Regression

## Objective:

Predict the median house value based on features like number of rooms, location, population, etc., using:

1 . Decision Tree Regressor

2 . Random Forest Regressor

3 . AdaBoost Regressor

1. 📊 Dataset

```python
from sklearn.datasets import fetch_california_housing
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = fetch_california_housing(as_frame= True)
```

```python
X = data.data
y = data.target
df = data.frame
```

```python
X.columns
```

```python
y
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state= 0 )
```

```python
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.ensemble import RandomForestRegressor, AdaBoostRegressor 
from  sklearn.tree import DecisionTreeRegressor
import numpy as np
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

Model 1 ---- Decision Tree

```python
tree = DecisionTreeRegressor(max_depth= 3)
tree.fit(X_train, y_train)
tree_preds = tree.predict(X_test)
```

Model 2----- RandomForest

```python

forest = RandomForestRegressor(n_estimators=100)
forest.fit(X_train, y_train)
forest_preds = forest.predict(X_test)

```

Model 3 -----AdaBoost

```python

ada = AdaBoostRegressor(n_estimators=100)
ada.fit(X_train, y_train)
ada_preds = ada.predict(X_test)
```

```python
def evaluate(model_name, y_true, y_pred):
    print(f"\n {model_name}")
    print(f"RMSE: {np.sqrt(mean_squared_error(y_true, y_pred)):.3f}")
    print(f"R2 Score: {r2_score(y_true, y_pred):.3f}")
evaluate("Decision Tree", y_test, tree_preds)
evaluate("Random Forest", y_test, forest_preds)
evaluate("AdaBoost", y_test, ada_preds)
```
