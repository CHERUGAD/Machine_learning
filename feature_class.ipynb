# Feature selection using F- classifire and Mutual_info_classifier

```python
import pandas as pd
```

```python
from sklearn.feature_selection import SelectKBest, f_classif, mutual_info_classif
```

```python
df = pd.read_csv('sample_dataset.csv')
df.columns
```

```python
df
```

```python
print(df.columns.tolist())
```

```python
df = df.loc[: ,['mean radius', 'mean texture', 'mean perimeter', 'mean area',
       'mean smoothness', 'mean compactness', 'mean concavity',
       'mean concave points', 'mean symmetry', 'mean fractal dimension',
       'target']].dropna()
```

```python
x = df.iloc[: , 0:10]
y = df['target']
y
```

```python
selector = SelectKBest(f_classif, k= 5)
selector.fit(x,y)
```

```python
selector.transform(x)
```

```python
get_featured = x.columns[selector.get_support()]
get_featured
```

## feature selection using mutual_info_classifier

```python
selector = SelectKBest(mutual_info_classif, k= 5)
selector.fit(x,y)
```

```python
selector.transform(x)
```

```python
get_featured = x.columns[selector.get_support()]
get_featured
```

### categorical feature and numerical target

```python
import pandas as pd
from sklearn.feature_selection import SelectKBest, mutual_info_regression
from sklearn.preprocessing import OrdinalEncoder,LabelEncoder
```

```python
df = pd.read_csv("sample_dataset_bins.csv")
df
```

```python
df['radius error'] = pd.cut(df['radius error'], 10)

```

```python
df
```

```python
x = df.iloc[: , 0:10]
y = df.iloc[:, -1]
x
```

```python
y
```

```python
x = OrdinalEncoder().fit_transform(x)
y = LabelEncoder().fit_transform(y)
```

```python
selector  = SelectKBest(lambda x,y : mutual_info_regression(x,y, discrete_features= True), k = 5 )
```

```python
selector.fit_transform(x,y)
```
