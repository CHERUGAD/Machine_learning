# Decision Tree 

Decision tree is a supervised machine learning algorithm can used in both classification and regression , datasets is splitted like branches and the decesion will be made and leaf node will store the output 

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
X = [[0, 1], [1,2]]
y = [0, 1]
```

```python
DTC = DecisionTreeClassifier()
DTC.fit(X, y)
```

```python
DTC.predict([[0,1]])
```

```python
DTC.predict([[2,2]])
```

```python
DTC.predict([[2.,1.]])
```

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
```

```python
iris  =  load_iris(as_frame = True)
```

```python
X = iris.data.iloc[:, 2:]
X
```

```python
y = iris.target
iris.feature_names
```

```python
y
```

```python
Dclf =  DecisionTreeClassifier()
```

```python
clf = Dclf.fit(X, y)
```

```python
clf
```

```python
from sklearn.tree import export_graphviz

from graphviz import Source

```

```python
dot_data = export_graphviz(clf,
                out_file= "tree.dot",
                feature_names= iris.feature_names[2:],
                class_names= iris.target_names,
                rounded = True,
                filled= True)
```

```python
dot_data
```
