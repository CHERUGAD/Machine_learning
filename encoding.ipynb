#Encoding of the categorical variables

#One hot encoding

```python
import numpy as np
from sklearn.preprocessing import OneHotEncoder
```

```python
x = np.array([['A'],['A'],['B'], ['c']])
x
```

```python
enc = OneHotEncoder()
```

```python
enc.fit_transform(x).toarray()
```

```python
enc = OneHotEncoder(sparse=False )
enc.fit_transform(x)
```

```python
enc = OneHotEncoder()
enc.fit_transform(x).todense()
```
