Ordinal Encoding


```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
x = [['High'], ['Low'], ['Medium'], ['Low']]
```

```python
ord_enc = OrdinalEncoder()
ord_enc.fit_transform(x)
```

We want 
!1)low ==0
2}medium ==1
high==2

```python
enc = OrdinalEncoder(categories=[['Low', 'Medium', 'High']])
enc.fit_transform(x)
```

```python
x = [['A', 'High'], ['B', 'Low'], ['C', 'Medium'], ['D', 'High']]
enc = OrdinalEncoder(categories=[['A', 'B', 'C', 'D'], ['Low', 'Medium', 'High']], handle_unknown='use_encoded_value', unknown_value=-1)
enc.fit_transform(x)
```

```python
from sklearn.preprocessing import LabelEncoder
x = ['A', 'B', 'C', 'D', 'A', 'B']
le = LabelEncoder()
le.fit_transform(x)
```

```python
from sklearn.preprocessing import OneHotEncoder
x = [['x','High'], ['y','Low'], ['z','Medium'], ['w','High']]
enc = OneHotEncoder(sparse=False, handle_unknown='ignore')
enc.fit_transform(x)
```

```python
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder
from sklearn.compose import ColumnTransformer
x = [['x','High'], ['y','Low'], ['z','Medium'], ['w','High']]
column_transform = ColumnTransformer(transformers=[
    ('onehot', OneHotEncoder(sparse_output=False, handle_unknown='ignore'), [0]),   # column 0
    ('ordinal', OrdinalEncoder(categories=[['High', 'Medium', 'Low']]), [1])        # column 1
])

# Transform the data
encoded = column_transform.fit_transform(x)
encoded
```
