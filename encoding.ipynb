```python
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder
from sklearn.compose import ColumnTransformer
x = [['x','High'], ['y','Low'], ['z','Medium'], ['w','High']]
column_transform = ColumnTransformer(transformers=[
    ('onehot', OneHotEncoder(sparse_output=False, handle_unknown='ignore'), [0]),   # column 0
    ('ordinal', OrdinalEncoder(categories=[['High', 'Medium', 'Low']]), [1])        # column 1
])

# Transform the data
encoded = column_transform.fit_transform(x)
encoded
```

```python
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
import pandas as pd

# Sample data
df = pd.DataFrame({'Size': ['Small', 'Medium', 'Large', 'Small', 'Large'],
                  'color': ['Red', 'Green', 'Blue', 'Green','Green']})

# Define transformer — using only OrdinalEncoder here because we only have one column
column_trans = ColumnTransformer(
    transformers=[
        ('ordinal', OrdinalEncoder(categories=[['Small', 'Medium', 'Large']]), ['Size']),
        ('onehot', OneHotEncoder(sparse_output =False, handle_unknown='ignore'), ['color'])
    ],
    remainder='passthrough'
)

# Apply transformation
encoded = column_trans.fit_transform(df)



print(encoded)
```
