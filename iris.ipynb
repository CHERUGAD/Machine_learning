```python
from sklearn.datasets import load_iris
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
iris  = load_iris()
```

```python
df = pd.DataFrame( data =  iris.data , columns= iris.feature_names)
df['target'] = iris.target
```

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
df1 = sns.load_dataset('iris')
```

```python
df1.head()
```

```python
df1.groupby('species').count()
```

```python
sns.pairplot(df1, hue= 'species', height = 3, aspect = 1)
plt.tight_layout()
```

```python
df1.hist( edgecolor = 'black', figsize= (12,5))

plt.show()
```

```python
numeric_cols = df1.select_dtypes(include='number').columns

# Create 2x2 subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 6))

# Flatten the axes array for easy indexing
axes = axes.flatten()

# Loop through each numeric column and plot histogram
for i, col in enumerate(numeric_cols):
    axes[i].hist(df1[col], edgecolor='black', linewidth=1.2, bins=15)
    axes[i].set_title(f' {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
