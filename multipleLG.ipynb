# Multiple Regression 

Multiple regression is a method which process one or more values as the input feature

```python
from sklearn.datasets import fetch_california_housing
import pandas as pd
```

```python
data = fetch_california_housing(as_frame= True)
```

```python
df = data.frame
```

```python
df
```

```python
df.shape
```

```python
x = df.iloc[:, :-1]
x
```

```python
y = data.target
y
```

# Stats model

stast model is a mathematical representation of  relationships between variales , used for understanding data, making predictions and testing hypothesis

```python
import statsmodels.api as sm
import statsmodels.formula as smf
```

```python
x_constatnt =  sm.add_constant(x)
```

```python
pd.DataFrame(x_constatnt)
```

```python
dir(sm.OLS)
```

```python
model = sm.OLS(y, x_constatnt).fit()
```

```python
model.summary()
```

```python
import matplotlib.pyplot as plt
```

```python
print(len(x))
print(len(y))
```

```python
y_prd = model.predict(x_constatnt)
```
