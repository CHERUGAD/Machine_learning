# Regression model with feature selection


```python
from sklearn.feature_selection import SelectKBest, f_regression, mutual_info_regression
from sklearn.linear_model import LinearRegression  
```

```python
x
```

```python
y
```

```python
testing = SelectKBest(f_regression, k = 1)
```

```python

testing.fit(x, y)
```

```python
x_tested = testing.transform(x)
```

```python
MODEL =LinearRegression()
MODEL.fit(x_tested, y)
```

```python
y_predicted  = MODEL.predict(x_tested)
```

```python
y_predicted.round(3)
```

```python
MODEL.intercept_
```

```python
MODEL.coef_
```

# Robust Regression

Robust regression is technique used to reduce the impact of outliers making the model more stable


```python
x = df[['RM']].values.reshape(-1, 1)
y = df['MEDV'].values
```

```python
from sklearn.linear_model import RANSACRegressor
```

```python
out_reducerer = RANSACRegressor()
out_reducerer.fit(x,y)
```

```python
inlier_mask = out_reducerer.inlier_mask_
outlier_mask = np.logical_not(inlier_mask)
```

```python
np.arange(3,10, 1)
```

```python
line_X = np.arange(3,10, 1)
line_Y_ransac = out_reducerer.predict(line_X.reshape(-1,1))
```

```python
sns.set(style= 'darkgrid', context= 'notebook')
plt.scatter(x[inlier_mask], y[inlier_mask], color='green',marker= 'o' ,label='Inliers')
plt.scatter(x[outlier_mask], y[outlier_mask], color='red',marker= 's', label='Outliers')
plt.plot(line_X, line_Y_ransac, color = 'black')
plt.legend()
plt.show()
```

```python
print(out_reducerer.estimator_.coef_)
out_reducerer.estimator_.intercept_
```
