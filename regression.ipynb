# Regression model with feature selection


```python
from sklearn.feature_selection import SelectKBest, f_regression, mutual_info_regression
from sklearn.linear_model import LinearRegression  
```

```python
df
```

```python
x = df[['RM']].values.reshape(-1, 1)
y = df['MEDV'].values
```

```python
x
```

```python
y
```

```python
testing = SelectKBest(f_regression, k = 1)
```

```python

testing.fit(x, y)
```

```python
x_tested = testing.transform(x)
```

```python
MODEL =LinearRegression()
MODEL.fit(x_tested, y)
```

```python
y_predicted  = MODEL.predict(x_tested)
```

```python
y_predicted.round(3)
```

```python
MODEL.intercept_
```

```python
MODEL.coef_
```

```python
plt.figure(figsize = (12,10))
sns.regplot(x= x,y=y)
plt.xlabel("Average number per dwelling ")
plt.ylabel("Median value according to rooms in 1000$")
plt.suptitle('LinearRegression used to predict ', color = 'orange' )
plt.title("LinearRegression used to predict ")
```

```python
plt.figure(figsize = (12,10))
sns.regplot(
    x=x_tested.flatten(), 
    y=y_predicted.flatten(), 
    scatter_kws={'color': 'black', 's': 40}, 
    line_kws={'color': 'blue', 'linewidth': 2}
)

plt.xlabel("Average number per dwelling ")
plt.ylabel("Median value according to rooms in 1000$")
plt.suptitle('LinearRegression used to predict ', color = 'orange' )
plt.title("LinearRegression used to predict ")
```

```python
x = x_tested.flatten()
y = y_predicted.flatten()
```

```python
plt.figure(figsize=(15, 6))
sns.jointplot(x =x, y= y, kind='reg', height= 10)
```
