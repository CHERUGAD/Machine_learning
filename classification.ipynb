# Performance measure and Stratified K fold

StratifiedKFold is a technique is used to ensure each fold has same proportion of classes as the original dataset 

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.base import clone
clf = SGDClassifier(random_state= 0)
```

```python
skfold  = StratifiedKFold(n_splits= 5)
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.base import clone

skfold = StratifiedKFold(n_splits=5)
accuracies = []

for train_inx, test_inx in skfold.split(x_train, y_train_0):
    clone_clf = clone(clf)
    x_train_fold = x_train.iloc[train_inx]
    y_train_fold = y_train_0.iloc[train_inx]
    x_test_fold  = x_train.iloc[test_inx]
    y_test_fold  = y_train_0.iloc[test_inx]

    clone_clf.fit(x_train_fold, y_train_fold)
    y_prd = clone_clf.predict(x_test_fold)
    n_correct = sum(y_prd == y_test_fold)

    accuracy = n_correct / len(y_prd)
    accuracies.append(accuracy)

    print(f"Fold accuracy: {accuracy:.4f}")

print(f"\nAverage accuracy across folds: {sum(accuracies) / len(accuracies):.4f}")
```

# Cross_val_score using KFold cross validation

```python
from sklearn.model_selection import cross_val_score
```

```python
scores = cross_val_score(clf, x_train, y_train_0, cv=4, scoring='accuracy')
print(scores)
print("Mean accuracy:", scores.mean())
```
