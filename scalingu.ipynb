# Scaling

Scaling is a data processing refers totransforming numerical features where it fall within a similar range.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

```python
df =  pd.read_csv ('sample_dataset.csv').iloc[:, 0 : 3].dropna()
df
```

## Normalization

0 - 1 scaling

```python
SCALER = MinMaxScaler()
x_scaled = SCALER.fit_transform(df)
x_scaled
```

```python
np.apply_over_axes(np.max,0, x_scaled)

```

```python
np.apply_over_axes(np.max,x_scaled, 0)
```

```python
np.apply_over_axes(np.min,x_scaled, 0)
```
