#Using column Transformer

```python
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
numerical =df.select_dtypes(include=['int64', 'float64']).columns
categorical = df.select_dtypes(include=['object', 'category']).columns
numerical
```

```python
categorical
```

```python
cleaner = ColumnTransformer(
    transformers=[
        ('numerical_transformer', SimpleImputer(strategy='mean'), numerical),
        ('categorical_transformer', SimpleImputer(strategy='most_frequent'), categorical)
        ])
```

```python
df6 = pd.DataFrame(cleaner.fit_transform(df), columns=numerical.tolist() + categorical.tolist())
df6
```

```python
cleaner = ColumnTransformer(
    transformers=[
        ('numerical_transformer', SimpleImputer(strategy='mean'), [0,1,2]),
        ('categorical_transformer', SimpleImputer(strategy='most_frequent'), categorical)
        ],remainder='passthrough')
df7 = pd.DataFrame(cleaner.fit_transform(df), columns=numerical.tolist() + categorical.tolist())
df7
```

Make column_selector


```python
from sklearn.compose import ColumnTransformer, make_column_selector
from sklearn.impute import SimpleImputer

cleaner = ColumnTransformer(
    transformers=[
        ('numerical_transformer', SimpleImputer(strategy='mean'),
         make_column_selector(dtype_include=['int64', 'float64'])),
        ('categorical_transformer', SimpleImputer(strategy='most_frequent'),
         make_column_selector(dtype_include=['object', 'category']))
    ],
    remainder='drop'
)
df8 = pd.DataFrame(cleaner.fit_transform(df), columns=numerical.tolist() + categorical.tolist())
df8
```
